```python
import kagglehub
path = kagglehub.dataset_download("shivamb/netflix-shows")
```

```python
from huggingface_hub import login
from google.colab import userdata

# Recupera el token de Hugging Face de los secretos de Colab
hf_token = userdata.get('HF_HUB_TOKEN')

# Autentica con el token
login(token=hf_token)

print("Autenticación con Hugging Face Hub completada.")
```

```python
from datasets import Dataset
import pandas as pd
import os

# Assuming 'path' variable from the previous cell holds the dataset directory
# Let's find the data file inside the downloaded path
data_file_path = os.path.join(path, "netflix_titles.csv") # Common filename for this dataset

# Load the data into a pandas DataFrame
df = pd.read_csv(data_file_path)

# Create a Hugging Face Dataset from the pandas DataFrame
dataset = Dataset.from_pandas(df)

# Push the dataset to the Hugging Face Hub
dataset.push_to_hub(
    "AngelGabrielTroncoso/mi_segundo_dataset"
)
```
