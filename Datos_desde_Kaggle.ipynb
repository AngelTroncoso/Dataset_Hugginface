```python
from huggingface_hub import login
from google.colab import userdata

# Recupera el token de Hugging Face de los secretos de Colab
hf_token = userdata.get('HF_HUB_TOKEN') # Corregido de 'HF_HUB_TOKEN' a 'HF_TOKEN'

# Autentica con el token
login(token=hf_token)

print("Autenticación con Hugging Face Hub completada.")
```

```python
from datasets import load_dataset, Dataset
import torch

# Define el ID del dataset de Hugging Face de origen
source_hf_dataset_id = "HuggingFaceH4/Multilingual-Thinking"

print(f"Cargando el dataset desde Hugging Face Hub: {source_hf_dataset_id}")

# --- Verificar disponibilidad de GPU ---
if torch.cuda.is_available():
    device = torch.device("cuda")
    print(f"¡GPU encontrada y disponible! Usando dispositivo: {torch.cuda.get_device_name(0)}")
else:
    device = torch.device("cpu")
    print("No se encontró GPU. Se utilizará la CPU.")
# --- Fin de la verificación de GPU ---

try:
    # Carga el dataset completo (sin streaming, asumiendo que es pequeño)
    dataset_dict = load_dataset(source_hf_dataset_id)
    print("Dataset cargado exitosamente. Características iniciales de los splits:")

    # Iterar sobre los splits para mostrar información
    for split_name, ds in dataset_dict.items():
        print(f"Split: {split_name}")
        print(f"  Número de ejemplos: {len(ds)}")
        print(f"  Características: {ds.features}")
        print("  Primeros 5 ejemplos:")
        # Dataset.head() es un método que devuelve un DataFrame de pandas para una vista previa
        display(ds.to_pandas().head(5))
        print("-" * 30)

    print("Dataset cargado y listo para ser explorado o procesado.")
    print("Puedes referirte a los splits como `dataset_dict['train']`, `dataset_dict['validation']`, etc.")

    # Define el nombre del nuevo repositorio en tu perfil de Hugging Face
    # Manteniendo el nombre anterior por si se desea subir este dataset
    target_hf_repo_name = "AngelGabrielTroncoso/tercer_dataset"
    print(f"El repositorio objetivo para subir es: {target_hf_repo_name}")

except Exception as e:
    print(f"¡Ocurrió un error inesperado durante la carga del dataset: {e}")
    print("Por favor, revisa el mensaje de error y la estructura del dataset.")
```
