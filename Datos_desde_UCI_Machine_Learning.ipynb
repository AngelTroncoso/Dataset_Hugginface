```python
from huggingface_hub import login
from google.colab import userdata

# Recupera el token de Hugging Face de los secretos de Colab
hf_token = userdata.get('HF_HUB_TOKEN')

# Autentica con el token
login(token=hf_token)

print("Autenticación con Hugging Face Hub completada.")
```

```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
heart_failure_clinical_records = fetch_ucirepo(id=519)

# data (as pandas dataframes)
X = heart_failure_clinical_records.data.features
y = heart_failure_clinical_records.data.targets

# metadata
print(heart_failure_clinical_records.metadata)

# variable information
print(heart_failure_clinical_records.variables)
```

```python
import pandas as pd
from datasets import Dataset

# Combine X and y into a single pandas DataFrame
df_combined = pd.concat([X, y], axis=1)

# Convert the pandas DataFrame to a Hugging Face Dataset
dataset = Dataset.from_pandas(df_combined)

# Define the name of the new repository in your profile of Hugging Face
target_hf_repo_name = "AngelGabrielTroncoso/cuarto_dataset"

print(f"Subiendo el dataset a tu espacio en Hugging Face Hub como: {target_hf_repo_name}")

# Sube el dataset a tu propio repositorio de Hugging Face
dataset.push_to_hub(target_hf_repo_name)

print(f"Dataset subido exitosamente a Hugging Face Hub: {target_hf_repo_name}")
```
