```python
from huggingface_hub import login
from google.colab import userdata

# Recupera el token de Hugging Face de los secretos de Colab
hf_token = userdata.get('HF_HUB_TOKEN')

# Autentica con el token
login(token=hf_token)

print("Autenticación con Hugging Face Hub completada.")
```

```python
from datasets import load_dataset, Dataset

# Define el ID del dataset de Hugging Face de origen
source_hf_dataset_id = "AlienKevin/SWE-ZERO-12M-trajectories"

print(f"Cargando el dataset desde Hugging Face Hub: {source_hf_dataset_id}")
# Carga el dataset
# Asegúrate de que el dataset es tabular. Si es un dataset de ficheros, la forma de cargarlo puede variar.
# Aquí asumimos que es un dataset con un splits (ej. 'train', 'validation')
# Si no tiene splits, podrías usar `dataset = load_dataset(source_hf_dataset_id)`
dataset = load_dataset(source_hf_dataset_id)

# Si el dataset tiene múltiples configuraciones o splits, puedes acceder a ellos así:
# Por ejemplo, si quieres el split 'train':
# train_dataset = dataset['train']

# Define el nombre del nuevo repositorio en tu perfil de Hugging Face
target_hf_repo_name = "AngelGabrielTroncoso/tercer_dataset"

print(f"Subiendo el dataset a tu espacio en Hugging Face Hub como: {target_hf_repo_name}")

# Sube el dataset cargado a tu propio repositorio de Hugging Face
# Si el dataset tiene splits (ej. dataset['train'], dataset['test']), necesitas subir cada split individualmente o iterar:
# Por simplicidad, si es un dataset con un solo split como 'train', puedes hacer:
# dataset['train'].push_to_hub(target_hf_repo_name)

# Si el objeto 'dataset' es un DatasetDict (contiene splits), puedes iterar sobre ellos
if isinstance(dataset, dict):
    for split_name, split_dataset in dataset.items():
        print(f"Subiendo split '{split_name}'...")
        split_dataset.push_to_hub(target_hf_repo_name, split=split_name)
else:
    # Si es un objeto Dataset directamente (sin splits iniciales como DatasetDict)
    dataset.push_to_hub(target_hf_repo_name)

print(f"Dataset subido exitosamente a Hugging Face Hub: {target_hf_repo_name}")
```
